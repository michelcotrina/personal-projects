```python
import time
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
from datetime import datetime

# 2. Obtener el mes actual
def obtener_mes_actual(driver):
    try:
        # Esperar hasta que el elemento esté presente
        mes_elemento = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.CLASS_NAME, "UyW9db"))
        )
        # Extraer, limpiar y devolver el texto del elemento
        mes_actual_texto = mes_elemento.text.strip()
        print(f"Mes actual extraído: '{mes_actual_texto}'")
        return mes_actual_texto
    except Exception as e:
        print(f"Error al obtener el mes actual: {e}")
        return None

# 3. Verificar si se llegó a la fecha límite
def verificar_fecha_limite(fecha_actual_texto, fecha_limite_texto):
    try:
        # Convertir ambos textos a objetos de fecha para comparación
        fecha_actual = datetime.strptime(fecha_actual_texto, "%B de %Y")
        fecha_limite = datetime.strptime(fecha_limite_texto, "%B de %Y")
        
        # Comparar las fechas
        if fecha_actual <= fecha_limite:
            print(f"Hemos llegado al mes límite: {fecha_limite_texto}.")
            return True
        return False
    except Exception as e:
        print(f"Error al verificar la fecha límite: {e}")
        return False

# 4. Extraer eventos del mes actual
def extraer_eventos_mes(driver):
    try:
        eventos = driver.find_elements(By.CLASS_NAME, "XuJrye")
        datos = []
        for evento in eventos:
            texto = evento.text
            if "Calendario:" in texto:
                partes = texto.split(", ")
                if len(partes) >= 4:
                    datos.append({
                        "Hora": partes[0],
                        "Descripción": partes[1],
                        "Calendario": partes[2].replace("Calendario: ", ""),
                        "Ubicación": partes[3].replace("Sin ubicación", "No especificado"),
                        "Fecha": partes[4] if len(partes) > 4 else "No especificada"
                    })
        print(f"Datos extraídos del mes actual: {len(datos)} eventos encontrados.")
        return datos
    except Exception as e:
        print(f"Error al extraer eventos: {e}")
        return []

# 5. Retroceder al mes anterior con espera
def retroceder_mes(driver):
    try:
        boton_anterior = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.XPATH, '//button[@aria-label="Mes anterior"]'))
        )
        boton_anterior.click()
        print("Retrocedimos al mes anterior.")
        # Esperar 15 segundos para que cargue la información
        time.sleep(8)
    except Exception as e:
        print(f"Error al retroceder al mes anterior: {e}")

# 6. Guardar datos incrementalmente en CSV
def guardar_datos_incrementales(datos, archivo_csv):
    try:
        # Crear un DataFrame con los datos nuevos
        data = pd.DataFrame(datos)
        # Guardar o agregar al archivo CSV
        if not data.empty:
            with open(archivo_csv, 'a', encoding='utf-8', newline='') as f:
                data.to_csv(f, index=False, header=f.tell() == 0)
            print(f"Datos guardados incrementalmente en '{archivo_csv}'.")
        else:
            print("No hay datos nuevos para guardar.")
    except Exception as e:
        print(f"Error al guardar los datos en el archivo CSV: {e}")

# 7. Iterar hasta la fecha límite y guardar en CSV
def iterar_hasta_fecha_limite(driver, fecha_limite_texto, archivo_csv):
    global todos_los_datos
    todos_los_datos = []  # Lista para acumular datos
    while True:
        try:
            # Obtener el mes actual
            fecha_actual_texto = obtener_mes_actual(driver)
            if fecha_actual_texto is None:
                print("Error al obtener el mes actual, terminando iteración.")
                break
            
            # Verificar si se llegó al límite
            if verificar_fecha_limite(fecha_actual_texto, fecha_limite_texto):
                print("Fecha límite alcanzada, deteniendo iteración.")
                break
            
            # Extraer datos del mes actual
            eventos_mes = extraer_eventos_mes(driver)
            todos_los_datos.extend(eventos_mes)  # Acumular datos
            
            # Guardar los datos extraídos de este mes
            guardar_datos_incrementales(eventos_mes, archivo_csv)
            
            # Retroceder al mes anterior
            retroceder_mes(driver)
        
        except KeyboardInterrupt:
            # Manejo de interrupción manual
            print("Interrupción manual detectada. Guardando datos acumulados y deteniendo...")
            guardar_datos_incrementales(todos_los_datos, archivo_csv)
            break
    
    # Mensaje final
    if todos_los_datos:
        print("Proceso finalizado. Datos recopilados correctamente.")
    else:
        print("No se encontraron datos para guardar.")

# 8. Fecha límite (ajustar según tu necesidad)
fecha_limite = "Diciembre de 2022"
archivo_csv = "agenda_presidencial.csv"

# 9. Ejecutar el proceso
iterar_hasta_fecha_limite(driver, fecha_limite, archivo_csv)
```

```python
def ordenar_y_guardar_csv(archivo_csv, columna_orden="Fecha"):
    try:
        # Leer el CSV en un DataFrame
        data = pd.read_csv(archivo_csv)
        # Ordenar el DataFrame por la columna de referencia
        data = data.sort_values(by=[columna_orden])
        # Guardar el DataFrame ordenado en el mismo archivo
        data.to_csv(archivo_csv, index=False)
        print(f"Datos ordenados por la columna '{columna_orden}' y guardados en '{archivo_csv}'.")
    except Exception as e:
        print(f"Error al ordenar y guardar el archivo CSV: {e}")
```

```python
import pandas as pd
import time
from datetime import datetime
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Variable global para rastrear el orden
contador = 0  

# 1. Obtener el mes actual
def obtener_mes_actual(driver):
    try:
        mes_elemento = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.CLASS_NAME, "UyW9db"))
        )
        mes_actual_texto = mes_elemento.text.strip()
        print(f"Mes actual extraído: '{mes_actual_texto}'")
        return mes_actual_texto
    except Exception as e:
        print(f"Error al obtener el mes actual: {e}")
        return None

# 2. Verificar si se llegó a la fecha límite
def verificar_fecha_limite(fecha_actual_texto, fecha_limite_texto):
    try:
        fecha_actual = datetime.strptime(fecha_actual_texto, "%B de %Y")
        fecha_limite = datetime.strptime(fecha_limite_texto, "%B de %Y")
        if fecha_actual <= fecha_limite:
            print(f"Hemos llegado al mes límite: {fecha_limite_texto}.")
            return True
        return False
    except Exception as e:
        print(f"Error al verificar la fecha límite: {e}")
        return False

# 3. Extraer eventos del mes actual
def extraer_eventos_mes(driver):
    try:
        eventos = driver.find_elements(By.CLASS_NAME, "XuJrye")
        datos = []
        for evento in eventos:
            texto = evento.text
            if "Calendario:" in texto:
                partes = texto.split(", ")
                if len(partes) >= 4:
                    datos.append({
                        "Hora": partes[0],
                        "Descripción": partes[1],
                        "Calendario": partes[2].replace("Calendario: ", ""),
                        "Ubicación": partes[3].replace("Sin ubicación", "No especificado"),
                        "Fecha": partes[4] if len(partes) > 4 else "No especificada"
                    })
        print(f"Datos extraídos del mes actual: {len(datos)} eventos encontrados.")
        return datos
    except Exception as e:
        print(f"Error al extraer eventos: {e}")
        return []

# 4. Retroceder al mes anterior con espera
def retroceder_mes(driver):
    try:
        boton_anterior = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.XPATH, '//button[@aria-label="Mes anterior"]'))
        )
        boton_anterior.click()
        print("Retrocedimos al mes anterior.")
        time.sleep(7)  # Esperar para que la página cargue completamente
    except Exception as e:
        print(f"Error al retroceder al mes anterior: {e}")

# 5. Guardar datos incrementalmente
def guardar_datos_incrementales(datos, archivo_csv):
    global contador
    try:
        # Agregar una columna "Orden" para rastrear el orden de extracción
        for evento in datos:
            evento["Orden"] = contador
            contador += 1
        # Crear un DataFrame con los datos nuevos
        data = pd.DataFrame(datos)
        # Guardar o agregar al archivo CSV
        if not data.empty:
            with open(archivo_csv, 'a', encoding='utf-8', newline='') as f:
                data.to_csv(f, index=False, header=f.tell() == 0)
            print(f"Datos guardados incrementalmente en '{archivo_csv}'.")
        else:
            print("No hay datos nuevos para guardar.")
    except Exception as e:
        print(f"Error al guardar los datos en el archivo CSV: {e}")

# 6. Ordenar y guardar el archivo CSV
def ordenar_y_guardar_csv(archivo_csv, columna_orden="Orden"):
    try:
        data = pd.read_csv(archivo_csv)
        data = data.sort_values(by=[columna_orden])
        data.to_csv(archivo_csv, index=False)
        print(f"Datos ordenados por la columna '{columna_orden}' y guardados en '{archivo_csv}'.")
    except Exception as e:
        print(f"Error al ordenar y guardar el archivo CSV: {e}")

# 7. Iterar hasta la fecha límite y guardar
def iterar_hasta_fecha_limite(driver, fecha_limite_texto, archivo_csv="agenda_presidencial.csv"):
    todos_los_datos = []
    try:
        while True:
            # Obtener el mes actual
            fecha_actual_texto = obtener_mes_actual(driver)
            if fecha_actual_texto is None:
                print("Error al obtener el mes actual, terminando iteración.")
                break
            
            # Verificar si se llegó al límite
            if verificar_fecha_limite(fecha_actual_texto, fecha_limite_texto):
                print("Fecha límite alcanzada, deteniendo iteración.")
                break
            
            # Extraer datos del mes actual
            eventos_mes = extraer_eventos_mes(driver)
            todos_los_datos.extend(eventos_mes)  # Acumular datos en memoria
            guardar_datos_incrementales(eventos_mes, archivo_csv)  # Guardar incrementalmente
            
            # Retroceder al mes anterior
            retroceder_mes(driver)
    except KeyboardInterrupt:
        print("\nProceso interrumpido manualmente.")
    finally:
        # Guardar y ordenar datos acumulados en caso de interrupción
        if todos_los_datos:
            ordenar_y_guardar_csv(archivo_csv)
        print("Proceso finalizado.")

# 8. Fecha límite (ajustar según tu necesidad)
fecha_limite = "Diciembre de 2022"

# 9. Ejecutar el proceso
iterar_hasta_fecha_limite(driver, fecha_limite)
```
